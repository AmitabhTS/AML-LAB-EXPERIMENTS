```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Correcting the separator to ';' so pandas can parse the columns correctly
marketing_df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/AML/marketing_campaign.csv", sep=";")

print("Shape:", marketing_df.shape)
display(marketing_df.head())

print("\nInformation:")
marketing_df.info()

print("\nSummary statistics:")
display(marketing_df.describe(include="all"))

print("\nMissing values:")
print(marketing_df.isnull().sum())

print("\nColumns:")
print(marketing_df.columns.tolist())
```

```python
target_marketing = "AcceptedCmp1"   # Change if necessary
```

```python
# Target distribution
plt.figure(figsize=(6, 4))
sns.countplot(
    data=marketing_df,
    x=target_marketing,
    hue=target_marketing,
    legend=False
)
plt.title("Marketing Campaign Response Distribution")
plt.show()
# Numerical distributions
numerical_columns = marketing_df.select_dtypes(
    include=["int64", "float64"]
).columns
marketing_df[numerical_columns].hist(
    figsize=(16, 12),
    bins=25
)
plt.tight_layout()
plt.show()
# Categorical columns
categorical_columns = marketing_df.select_dtypes(
    include=["object", "category"]
).columns

print("Numerical features:", list(numerical_columns))
print("Categorical features:", list(categorical_columns))
```
