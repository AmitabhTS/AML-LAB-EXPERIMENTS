```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# The file appears to be comma-separated, so removing the 'sep=;' argument
# will allow pd.read_csv to use its default comma delimiter.
car_df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/AML/CarPrice.csv")

print("Shape:", car_df.shape)
display(car_df.head())

print("\nInformation:")
print(car_df.info())

print("\nMissing values:")
print(car_df.isnull().sum())

print("\nDescriptive statistics:")
display(car_df.describe(include="all"))

print("\nColumn names:")
print(car_df.columns.tolist())

target_car = "price"       # Change to "Price" or "selling_price" if required

plt.figure(figsize=(8, 5))
sns.histplot(car_df[target_car])
plt.title("Distribution of Used-Car Prices")
plt.show()

categorical_columns = car_df.select_dtypes(
    include=["object", "category"]
).columns

print("Categorical features:")
print(list(categorical_columns))

numerical_columns = car_df.select_dtypes(
    include=["int64", "float64"]
).columns

print("Numerical features:")
print(list(numerical_columns))

# Price by fuel type, if available
if "fuel" in car_df.columns:
    plt.figure(figsize=(8, 5))
    sns.boxplot(data=car_df, x="fuel", y=target_car, hue="fuel", legend=False)
    plt.xticks(rotation=30)
    plt.title("Car Price by Fuel Type")
    plt.show()

# Price by transmission, if available
if "transmission" in car_df.columns:
    plt.figure(figsize=(7, 5))
    sns.boxplot(
        data=car_df,
        x="transmission",
        y=target_car,
        hue="transmission",
        legend=False
    )
    plt.title("Car Price by Transmission")
    plt.show()

# Correlation heatmap
plt.figure(figsize=(10, 7))
corr = car_df.select_dtypes(include="number").corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix of Numerical Car Features")
plt.show()
```
