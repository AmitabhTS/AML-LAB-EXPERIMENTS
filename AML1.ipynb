```python
# Experiment 1:
# Understanding "Mobile Price" Dataset by Doing Feature Analysis

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --------------------------------------------------
# 1. Load the dataset
# --------------------------------------------------

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/train.csv")

print("Dataset loaded successfully!")

# --------------------------------------------------
# 2. Display first five records
# --------------------------------------------------

print("\nFirst five records:")
print(df.head())

# --------------------------------------------------
# 3. Display dataset dimensions
# --------------------------------------------------

print("\nNumber of rows and columns:")
print(df.shape)

# --------------------------------------------------
# 4. Display feature names
# --------------------------------------------------

print("\nFeatures in the dataset:")
print(df.columns.tolist())

# --------------------------------------------------
# 5. Display dataset information
# --------------------------------------------------

print("\nDataset Information:")
df.info()

# --------------------------------------------------
# 6. Statistical description
# --------------------------------------------------

print("\nStatistical Summary:")
print(df.describe())

# --------------------------------------------------
# 7. Check data types
# --------------------------------------------------

print("\nData Types:")
print(df.dtypes)

# --------------------------------------------------
# 8. Check target variable
# --------------------------------------------------

print("\nPrice Range Distribution:")
print(df[['price_range']].value_counts().sort_index())

# --------------------------------------------------
# 9. Visualize target variable
# --------------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='price_range')

plt.title("Distribution of Mobile Price Range")
plt.xlabel("Price Range")
plt.ylabel("Number of Mobiles")

plt.show()

# --------------------------------------------------
# 10. Distribution of numerical features
# --------------------------------------------------

df.hist(figsize=(18, 15), bins=20)

plt.suptitle("Distribution of Mobile Price Dataset Features")
plt.tight_layout()

plt.show()

# --------------------------------------------------
# 11. Correlation matrix
# --------------------------------------------------

correlation = df.corr(numeric_only=True)

print("\nCorrelation with Price Range:")
print(correlation['price_range'].sort_values(ascending=False))

# --------------------------------------------------
# 12. Correlation heatmap
# --------------------------------------------------

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()

# --------------------------------------------------
# 13. Feature correlation with price range
# --------------------------------------------------

price_corr = correlation['price_range'].drop('price_range')
price_corr = price_corr.sort_values()

plt.figure(figsize=(10, 7))

price_corr.plot(kind='barh')

plt.title("Feature Correlation with Price Range")
plt.xlabel("Correlation")
plt.ylabel("Features")

plt.show()

# --------------------------------------------------
# 14. Important feature analysis
# --------------------------------------------------

important_features = [
    'battery_power',
    'int_memory',
    'px_height',
    'px_width',
    'ram'
]

for feature in important_features:

    plt.figure(figsize=(7, 5))

    sns.boxplot(
        data=df,
        x='price_range',
        y=feature
    )

    plt.title(f"{feature} vs Price Range")
    plt.xlabel("Price Range")
    plt.ylabel(feature)

    plt.show()

print("\nFeature analysis completed successfully.")
```
